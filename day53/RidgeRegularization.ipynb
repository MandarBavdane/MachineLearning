```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes

data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
X = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
LR = LinearRegression()
```

```python
LR.fit(X_train, y_train)
```

```python
y_pred = LR.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error

print('R2 score:', r2_score(y_test, y_pred))
print('RMSE:', np.sqrt(mean_squared_error(y_test, y_pred)))
```

```python
from sklearn.linear_model import Ridge
R = Ridge(alpha = 0.0001) # Lambda in our case, default: 1
```

```python
R.fit(X_train, y_train)
```

```python
y_pred1 = R.predict(X_test)
```

```python
print('R2 Score:', r2_score(y_test, y_pred1))
print("RMSE:",  np.sqrt(mean_squared_error(y_test, y_pred1)))
```

```python
# Generating Non-linear Dataset
m = 100
x1 = 5 * np.random.rand(m, 1) - 2 # Uniform Distribution [0, 1] +ve values only
x2 = 0.7 * x1**2 - 2*x1 + 3 + np.random.randn(m, 1) # Normal(Gaussian) Distribution () (-∞, +∞) (can be +ve or -ve)

plt.scatter(x1, x2)
plt.show
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)), # polynomial_reg of deg=16
        ('ridge', Ridge(alpha=alpha)) # solveed using ridge
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200] # lambdas are 0, 20, 200
cs = ['r', 'g', 'b'] # There 0 will perform as LR

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```
